```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn import svm
```

Generate a toy data sample for our classification purpose

```python
# The data is 2Dimesnonal.
x= np.array([1, 2, 5, 1.5, 8, 1, 9, 7, 8.7, 2.3, 5.5, 7.7, 6.1, 8, 8])
y= np.array([2, 4, 8, 1.8, 8, 0.6, 11, 10, 9.4, 4, 3, 8.8, 7.5, 8, 2])
print(x)
print(y)
```

Let's visualize how the data looks.

```python
plt.scatter(x,y)
plt.show()
```

Class labels are added to each sample so that it remains linearly separable.

```python
# Combine x and y arrays into a 2D feature matrix
training_X = np.column_stack((x, y))

# Create binary class labels that are linearly separable
# For example, class 0 if y < x, else class 1
training_y = np.where(y < x, 0, 1)

# Define and fit the support vector classifier
clf = svm.SVC(kernel='linear', C=1.0)
clf.fit(training_X, training_y)
```

**In a Jupyter environment, please rerun this cell to show the HTML representation or trust the notebook.**

The trained model predicts the class and draws the decision boundary.

```python
# Extract the model parameter for the linear SVM from the trained SVM model
w = clf.coef_[0]

# Calculate the y-offset for the linear SVM
a = -w[0] / w[1]

# make the x-axis space for the data points
XX = np.linspace(0, 13)

# get the labels to plot the decision boundary
yy = a * XX - clf.intercept_[0] / w[1]

# plotting the decision boundary
plt.plot(XX, yy, 'k-')

# showing the plot visually
plt.scatter(training_X[:, 0], training_X[:, 1], c=training_y)
plt.legend()
plt.show()
```
